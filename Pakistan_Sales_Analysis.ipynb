## **Data Cleaning**

```python
#import pyhton libraries

import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
%matplotlib inline              
import seaborn as sns
```

```python
# import csv file
df = pd.read_csv('Pakistan Sales Data.csv', encoding= 'unicode_escape')
```

```python
df.shape
```

```python
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
#drop unrelated/blank columns
df.drop(['Status', 'unnamed1'], axis=1, inplace=True)
```

```python
pd.isnull(df)
```

```python
pd.isnull(df).sum()
```

```python
# drop null values
df.dropna(inplace=True)  #inplace=true is use for save the changes
```

```python
# change data type
df['Amount'] = df['Amount'].astype('int') 
```

```python
df['Amount'].dtypes
```

```python
df.columns
```

```python
# describe() method returns description of the data in the DataFrame (i.e. count, mean, std, etc)
df.describe()
```

```python
# use describe() for specific columns
df[['Age', 'Orders', 'Amount']].describe()
```

## **Expolatory Data Analysis**

```python
df.columns
```

### **Gender**

```python
# plotting a bar chart for Gender and it's count

ax = sns.countplot(x = 'Gender', data = df,hue='Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
```

```python
sales_gen
```

```python
sns.barplot(x = 'Gender',y= 'Amount' , data = sales_gen,hue='Gender')
```

 ##### *From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men.*

### **Age**

```python
ax = sns.countplot(x = 'Age Group',data = df,  hue = 'Gender')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total Amount vs Age Group
sales_age = df.groupby(['Age Group'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

ax=sns.barplot(x = 'Age Group',y= 'Amount' ,data = sales_age,hue='Age Group')
```

##### *From above graphs we can see that most of the buyers are of age group between 26-35 yrs female.*

### **State**

```python
# total number of orders from states

sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False)
```

```python
sales_state
```

```python
sns.set(rc={'figure.figsize':(16,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Orders', hue='State')
```

```python
# total amount/sales from states

sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sales_state
```

```python
sns.set(rc={'figure.figsize':(16,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Amount',hue='State')
```

### **Marital Status**

```python
ax = sns.countplot(data = df, x = 'Marital_Status',hue='Marital_Status')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_marital= df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sales_marital
```

```python
sns.barplot(data = sales_marital, x = 'Marital_Status',y= 'Amount', hue='Gender')
```

##### *From above graphs we can see that most of the buyers are married (women) and they have high purchasing power*

### **Occupation**

```python
sns.set(rc={'figure.figsize':(18,6)})
ax = sns.countplot(data = df, x = 'Occupation',hue='Occupation')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_occupation = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sales_occupation
```

```python
sns.set(rc={'figure.figsize':(18,5)})
sns.barplot(data = sales_occupation, x = 'Occupation',y= 'Amount',hue='Occupation')
```

### **Product Category**

```python
sns.set(rc={'figure.figsize':(25,5)})
ax = sns.countplot(x = 'Product_Category',data = df, hue='Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
#plotting grapgh for top 10 product
sales_product = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_product, x = 'Product_Category',y= 'Amount',hue='Product_Category')
```

##### *From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category*

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders',hue='Product_ID')
```

### **Conclusion**

###### *Married women age group 26-35 yrs from Sindh, Balochistan and Punjab are working in IT, Textile and Retail are more likely to buy products from Food, Clothing and Electronics category.*
